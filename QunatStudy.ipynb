# ***Основные термины***

bid - цена покупателя (за такую цену готовы купить акцию)

ask - цена продавца (за такую цену готовы продать акцию)

Стакан - это сбор всех bid и ask цен и выглядит он вот так:

bid........цена...........ask

500......99.80$

300......99.90$

.............100.10$......300

.............100.20$......500

Тут показано как выглядит стакан, написано сколько акций готовы купить или продать и за какую цену.


*Спред* - это разница между ценой покупки и продажи. На примере выше, спред равен 0.2$

***Ордеры***

Ордер - это по сути заказ на бирже. Он бывает лимитным и рыночным.

Рыночный - покупаешь или продаешь по рынку, прямо сейчас.

Лимитный - покупаешь по своей цене, но только если найдется продавец или покупатель (если цена дойдет до твоего ордера). Лимитный ордер может никогда не выполнится

***Ликвидность***

Есть несколько понятий, напишу пока только одно. Ликвидность товара, акции, монеты, актива - насколько быстро его можно продать.

***Слиппаж***

Ты отправил ордер на покупку акции за 100, но цена резко выросла и ты купил за 100.05

0.05 - это и есть слиппаж, он может быть в обе стороны. Такое происходит, если: рынок быстро двигается (волатильный), ты сделал большой ордер и ты просто выкупил все по цене 100, ликвидность низкая

# **Альфа и Бета**

*Альфа* - это доходность сверх рынка, т.е. например рынок вырос на 67%, а ты заработал 69%, значит твоя альфа 2%
*Бета* - это насколько твоя стратегия повторяет рынок, т.е. если твоя бета = 2.0, рынок вырос на 10%, ты заработаешь 20%

Доходность можно посчитать так

Доходность = а + б * доходность рынка, т.е. например у тебя альфа = 3, бета = 1.2, рынок вырос на 10%:

3 + 1.2 * 10 = 3 + 12 = 15% - вот столько ты заработаешь при таких параметрах, здесь 12% из-за того что вырос рынок, а 3 из-за твоей стратегии

# **Распределение**

На рынке есть такое понятие распределение, если по простому, то как в среднем ведет себя рынок, математики думают, что варьируется от -3% до +3% за день. Большие значение почти никогда не случаются. Но прикол в том, что это не так.

Рынки падали (2008 год -22% за день) и вот такое математики не учитывают, из-за чего можно потерять все деньги

*Что это значит?*

Надо предусматривать такой вариант, что рынки могут обвалится, могут опуститься на -10%+ за день, т.е. защищать капитал от такого fat tails

Также рынок ассиметричен это значит, что он движется не одинаково постоянно и неравномерно. Если упали на -10, это не значит, что мы отрастем за такой же срок на +10%



> 🧠 Большинство акций имеют отрицательный skew. Ты долго зарабатываешь по чуть-чуть — потом раз, и большой минус.





# **Станционарность**

Станционарность - это когда что-то постоянно на одном уровне, т.е. +- постоянно. Акции обычно не станционарны. А доходность станционарна.

Акция растет каждый день в два раза: 100 -> 200 -> 400 -> 800
Цена точно нестанционарна
Доходность: 100% -> 100% -> 100% -> 100%
Она станциорна, поэтому стоит опираться на доходность, а не на цену

**ADF тест** - это тест, станционарна акция или нет (есть тренд или нет).
ΔPt​=α+ρ⋅Pt−1​+εt - формула ADF

Страшно выглядит — давай переведём каждый кусок:

ΔP_t — насколько цена изменилась сегодня

P_{t-1} — цена вчера

ρ (ро) — вот это и есть главный коэффициент

ε_t — случайный шум


В конце выдается одно число и по нему уже нужно судить, есть тренд или нет.

<0.05 - станционарен (нет тренда)
>0.05 - нестанционарен (есть тренд)

1.   Не всегда видно, есть ли тренд(бывает не заметный)

# **Sharpe Ratio**

Не всегда чисто доходность делает стратегию лучше, еще нужно знать, какой был риск - это и есть sharpe

Sharpe = (R_p - R_f)/σ - формула

R_p — доходность стратегии
Сколько заработала твоя стратегия за год. Например +20%.

R_f — безрисковая ставка
Сколько ты заработал бы вообще без риска — просто положив деньги в банк или купив государственные облигации. Например +4%.

σ (сигма) — стандартное отклонение доходностей
Насколько сильно скакала доходность каждый день. Это и есть риск.

Чтоб объяснить стандартное отклонение приведу пример, где 2 трейдера зарабатывают +1% в день в среднем:

Трейдер А — доходности по дням:

+1%, +1.1%, +0.9%, +1%, +1% ...

Разброс маленький — σ = 0.1%

Трейдер Б — доходности по дням:

+5%, -3%, +4%, -2%, +3% ...

Разброс огромный — σ = 4%

**Sortino Ratio** - умный брат Sharpe Ratio

У Sharpe есть проблема, он штрафует за любую волатильность, даже хорошую. Например ты сделал за день +10%, но Sharpe скажет, что это плохо.

Sortino считает только убыточную волатильность, т.е. в формулу вместо σ, ставиться σ_downside - стандартное отклонение только убыточных дней

Из-за этого Sortino всегда будет выше Sharpe

Обычно считают оба и сравнивают, если оба высокие >2, то тогда стратегия хорошая, но если один или оба <2, то стоит пересмотреть свою стретегию.



> 🧠 Если Sortino сильно выше Sharpe — значит стратегия делает редкие но огромные прибыли. Это сигнал разобраться откуда они берутся.



# **Drawdown**

Drawdown - это реальный риск стратегии.

Наглядный пример, за что отвечает эта метрика:

Представь что ты вложил 10,000$ в стратегию. Вот что происходило:

10,000$ → 13,000$ → 14,000$ → 9,000$ → 11,000$ → 15,000$

Sharpe и Sortino говорят тебе "в среднем всё хорошо".

Но посмотри на момент когда было 14,000$ → стало 9,000$. Ты потерял 5,000$ от пика. Это -35.7%.

Как ты себя чувствуешь в этот момент? Ты готов держать стратегию дальше или запаникуешь и выйдешь?

Drawdown — это и есть ответ на вопрос: "Насколько больно будет держать эту стратегию?"

Drawdown = (Капитал - Пик) ​/ Пик - формула для расчета каждого дня

пик - пик цены до сегодня (включая)

капитал - баланс сегодня

 *Maximum Drawdown* = самое плохое значение

 ***Calmar Ratio*** — доходность за риск через MDD

Sharpe делит доходность на волатильность. Calmar делит доходность на максимальный drawdown

Пример:

Годовая доходность = +25%
MDD = -18.4%

Calmar = 25% / 18.4% = 1.36

***Win Rate и Profit Factor***

Win Rate - насколько часто у тебя прибыльные сделки(простая средняя)

Но Win Rate не всегда показывает картину целиком, ты можешь постоянно получать маленькую прибыль, а за одну убыточную сделку потерять все, win rate будет 90%, но по факту у тебя убытки

Поэтому Profit Factor показывает твою доходность по сделкам (метрика)

ProfitFactor = Сумма всех убытков / Сумма всех прибылей​

Пример с числами:
Сделки: +2%, -1%, +3%, -2%, +4%, -1%

Сумма прибылей: 2 + 3 + 4 = 9%
Сумма убытков: 1 + 2 + 1 = 4%

ProfitFactor = 4/9 = 2.25

# **Return**

Есть два типа: *Simple Return* и *Log Return*

Как работают return:

Simple:

```python
import pandas as pd

prices = pd.Series([100, 102, 99, 104, 101])
simple_returns = prices.pct_change().dropna()
print(simple_returns)
# 0.0200, -0.0294, 0.0505, -0.0288
```

Т.е. была цена 100$, на след день стала 102$ => 102 - 100 = 2 => 2%
и так каждый следующий день

Log:

```python
import numpy as np

log_returns = np.log(prices / prices.shift(1)).dropna()
print(log_returns)
# 0.0198, -0.0299, 0.0493, -0.0292
```

Здесь логика та же, но считается по другому, через логарифм, разницу в первом дне мы логарифмуем в ln(1.02) = 19.8%

Разница небольшая, но есть ключевые особенности, почему стоит использовать именно Log в коде, а Simple для публики, отчетов:

Если нужно посчитать доходность за какой-то период времени, то если мы будем просто складывать проценты Simple, то это неправильно, пример:
100 -> 50% -> -50%, если просто сложим проценты, то получим 0 и вроде останется 100$, но на самом деле у нас получилось 100 -> 150 -> 75, мы просто потеряли 25$

Если же Log, то будет так ln(1.50)+ln(0.50) = 0.405+(−0.693)= −0.288

e^−0.288 = 75$, у нас получилас реальная цена

Если нужно работать с временным рядом используем Log, для отчетов и простоты Simple

***Cummulative Return*** = это сложение доходности, чтоб получить итоговую доходность формула: (1 + (%)) * n, где n - это торговые дни представленные в виде (1 + (%)), % - доходность за день



***Ануализация*** - это средняя дневная доходность
формула: r = сумма всех % / кол-во дней
Зная среднюю доходность можно узнать среднюю доходность за год: r * 252 -  средняя годовоя доходность

# Метрики, но кодом

```python
import numpy as np
import pandas as pd

# Equity curve стратегии
capital = np.array([10000, 11000, 12500, 11800, 10200, 13000, 12000, 15000])

# Дневные доходности
returns = np.array([0.02, -0.01, 0.03, -0.005, 0.015, -0.01, 0.025, 0.01, -0.02, 0.03])

# Безрисковая ставка — 4% годовых → переводим в дневную, R_f
risk_free_annual = 0.04
risk_free_daily = risk_free_annual / 252
print(f"Дневная безрисковая ставка: {risk_free_daily:.6f}")
# 0.000159
```

```python
#Сейчас пишем Sharpe Ratio, формула Sharpe = (R_p - R_f)/σ

#Средняя дневная доходность R_p
mean_return = np.mean(returns)
print(f"Средняя дневная доходность: {mean_return:.4f}")

#Стандартное отклонение
std_return = np.std(returns)
print(f"Стандартное отклонение: {std_return:.4f}")

#Sharpe дневной
sharpe_daily = (mean_return - risk_free_daily) / std_return
print(f"Sharpe дневной: {sharpe_daily:.4f}")

#Sharpe годовой
sharpe_annual = sharpe_daily * np.sqrt(252)
print(f"Sharpe годовой: {sharpe_annual:.4f}")
```

```python
#Пишем тоже самое, но теперь Sortino

#Берем только отрицательные доходности
negative_returns = returns[returns < 0]
print(f"Убыточные дни: {negative_returns}")

#Стандартное отклонение от убытков
downside_std = np.std(negative_returns)
print(f"Downside std: {downside_std:.4f}")

#Sortino
sortino = (mean_return - risk_free_daily)/ downside_std * np.sqrt(252)
print(f"Годовой Sortino: {sortino:.4f}")
```

```python
#Maximum Drawdown
# Шаг 1 — строим equity curve из доходностей
equity_curve = (1 + returns).cumprod() * 10000
print(f"Equity curve:\n{np.round(equity_curve, 2)}")
# [10200. 10098. 10401.  10349. 10505. 10400. 10660. 10767. 10551. 10868.]

# Шаг 2 — пик на каждый момент времени
rolling_max = np.maximum.accumulate(equity_curve)
print(f"Пики:\n{np.round(rolling_max, 2)}")
# [10200. 10200. 10401. 10401. 10505. 10505. 10660. 10767. 10767. 10868.]

# Шаг 3 — drawdown каждого дня
drawdown = (equity_curve - rolling_max) / rolling_max
print(f"Drawdowns:\n{np.round(drawdown, 4)}")
# [ 0.    -0.01   0.    -0.005  0.    -0.01   0.     0.    -0.02   0.  ]

# Maximum Drawdown — самое плохое значение
max_drawdown = np.min(drawdown)
print(f"Maximum Drawdown: {max_drawdown:.2%}")
# -2.00%
```

```python
#Calmar Ratio
# Годовая доходность
annual_return = (1 + mean_return) ** 252 - 1
print(f"Годовая доходность: {annual_return:.2%}")
# 1174.23% (данные очень хорошие — в реальности так не бывает)

# Calmar
calmar = annual_return / abs(max_drawdown)
print(f"Calmar Ratio: {calmar:.2f}")
```

```python
# Win Rate
winning_days = returns[returns > 0]
win_rate = len(winning_days) / len(returns)
print(f"Win Rate: {win_rate:.2%}")
# 60.00%

# Profit Factor
total_profit = np.sum(returns[returns > 0])
total_loss = abs(np.sum(returns[returns < 0]))

profit_factor = total_profit / total_loss
print(f"Total profit: {total_profit:.4f}")
print(f"Total loss:   {total_loss:.4f}")
print(f"Profit Factor: {profit_factor:.2f}")
# Profit Factor: 2.25
```

```python
#Все метрики, но в одной функции
def strategy_metrics(returns, risk_free_annual=0.04):
    """
    Считает все основные метрики стратегии.

    returns — массив дневных доходностей
    risk_free_annual — безрисковая ставка годовая
    """
    rf_daily = risk_free_annual / 252

    # Базовые
    mean_r = np.mean(returns)
    std_r = np.std(returns)

    # Sharpe
    sharpe = (mean_r - rf_daily) / std_r * np.sqrt(252)

    # Sortino
    neg_returns = returns[returns < 0]
    downside = np.std(neg_returns)
    sortino = (mean_r - rf_daily) / downside * np.sqrt(252)

    # Max Drawdown
    equity = (1 + returns).cumprod()
    peak = np.maximum.accumulate(equity)
    dd = (equity - peak) / peak
    mdd = np.min(dd)

    # Calmar
    annual_r = (1 + mean_r) ** 252 - 1
    calmar = annual_r / abs(mdd)

    # Win Rate и Profit Factor
    win_rate = len(returns[returns > 0]) / len(returns)
    profit_factor = (np.sum(returns[returns > 0]) /
                     abs(np.sum(returns[returns < 0])))

    print("=" * 35)
    print("   МЕТРИКИ СТРАТЕГИИ")
    print("=" * 35)
    print(f"Годовая доходность : {annual_r:.2%}")
    print(f"Sharpe Ratio       : {sharpe:.2f}")
    print(f"Sortino Ratio      : {sortino:.2f}")
    print(f"Max Drawdown       : {mdd:.2%}")
    print(f"Calmar Ratio       : {calmar:.2f}")
    print(f"Win Rate           : {win_rate:.2%}")
    print(f"Profit Factor      : {profit_factor:.2f}")
    print("=" * 35)

# Запускаем
strategy_metrics(returns)
```

# Базовый бектест кодом

```python
import pandas as pd
import numpy as np

# Цены за 10 дней
data = pd.DataFrame({
    'price': [100, 102, 99, 104, 101, 106, 103, 108, 105, 110]
})

#скользящая средняя за 3 дня
data['ma3'] = data['price'].rolling(window=3).mean()
print(data)

print()
#был сигнал или нет
data['signal'] = 0
data.loc[data['price'] > data['ma3'], 'signal'] = 1
print(data)

print()
#Доходность каждый день
data['market_return'] = data['price'].pct_change()
#Сигнал сегодня -> позиция завтра
data['position'] = data['signal'].shift(1)
#Доходность стратегии
data['strategy_return'] = data['position'] * data['market_return']

print(data[['price', 'signal', 'position', 'market_return', 'strategy_return']].round(4))
print()

commission = 0.001

#Вход/выход из позиции, платим комсу
data['trade'] = data['position'].diff().abs()
data['commission'] = data['trade'] * commission

#В итоге какая доходность
data['net_return'] = data['strategy_return'] - data['commission']
print(data[['position', 'strategy_return', 'commission', 'net_return']].round(4))
print()
```

```python
#Строим график и считаем итог

clean = data.dropna()
market_equity = (1 + clean['market_return']).cumprod() * 10000
strategy_equity = (1 + clean['net_return']).cumprod() * 10000

strategy_returns = clean['net_return'].values
sharpe = (np.mean(strategy_returns) / np.std(strategy_returns)) * np.sqrt(252)
peak = strategy_equity.cummax()
mdd = ((strategy_equity - peak) / peak).min()
print(f"Sharpe:        {sharpe:.2f}")
print(f"Max Drawdown:  {mdd:.2%}")
print(f"Итог стратегии: {strategy_equity.iloc[-1]:.0f}$")
print(f"Итог рынка:     {market_equity.iloc[-1]:.0f}$")

# График
plt.figure(figsize=(10, 5))
plt.plot(market_equity.values, label='Рынок',
         color='gray', linestyle='--')
plt.plot(strategy_equity.values, label='Стратегия',
         color='blue', linewidth=2)
plt.axhline(10000, color='black', linewidth=0.8)
plt.title('Стратегия vs Рынок')
plt.ylabel('Капитал ($)')
plt.legend()
plt.tight_layout()
plt.show()
```

***Скелет, любого бектеста***
1. Данные     → price
2. Индикатор  → ma3 = rolling(3).mean()
3. Сигнал     → price > ma3 → signal = 1
4. Позиция    → signal.shift(1)
5. Доходность → position × market_return
6. Комиссия   → diff().abs() × 0.001
7. Метрики    → Sharpe, MDD, equity curve

Это скелет который работает для любой стратегии. Меняем только шаг 2 и 3 — всё остальное одинаково.

# Walk-Forward Validation


Если просто, то это защита от переобучения только в Quant.

Если наглядно

Обучение на всех данных(неправильно):

2021 ████████████████████████

2022 ████████████████████████

2023 ████████████████████████

     ↑
     Стратегия видела всё → результаты врут

Walk-forward (правильно):

Шаг 1:

  Обучение:  2021 ████████████

  Проверка:  2022             ░░░░░░░░

Шаг 2:

  Обучение:  2021-2022 ████████████████████

  Проверка:  2023                          ░░░░░░░░

Результат = только проверочные периоды ░░░░░░░░



> 🧠 Стратегия никогда не видит данные на которых проверяется. Это честный результат.





```python
#Пишем walk-forward

import pandas as pd
import numpy as np
from sklearn.model_selection import TimeSeriesSplit

#генерация данных
np.random.seed(42)
dates = pd.date_range('2021-01-01', '2023-12-31', freq='B')
price = pd.Series(
    100 * (1 + np.random.randn(len(dates)) *  0.01).cumprod(),
    index=dates,
    name='price'
)

print(f'Всего дней: {len(price)}')

# TimeSeriesSplit — специальный сплиттер для временных рядов
# Обычный train_test_split НЕЛЬЗЯ использовать — он перемешивает данные

tscv = TimeSeriesSplit(n_splits=3, test_size=126)
for fold, (train_idx, test_idx) in enumerate(tscv.split(price)):
  train = price.iloc[train_idx]
  test = price.iloc[test_idx]

  #обучение
  print(f"Фолд {fold + 1}:")
  print(f"  Обучение: {train.index[0].date()} → {train.index[-1].date()} ({len(train)} дней)")
  print(f"  Проверка: {test.index[0].date()} → {test.index[-1].date()} ({len(test)} дней)")
  print()


def run_strategy(price, window):
  """Простая MA стратегия. Возвращает Sharpe."""
  ma = price.rolling(window).mean()
  signal = (price > ma).shift(1).fillna(0)

  returns = price.pct_change() * signal
  returns = returns.dropna()

  if returns.std() == 0:
    return 0

  sharpe = returns.mean() / returns.std() * np.sqrt(252)
  return sharpe

def find_best_window(train_price, window=[3, 5, 10, 20]):
  """Находим лучший параметр на train данных."""
  results = {}
  for w in window:
    results[w] = run_strategy(train_price, w)

  best_window = max(results, key=results.get)
  return best_window, results


#Запускаем walk-Forward
wf_results = []

tscv = TimeSeriesSplit(n_splits=3)

for fold, (train_idx, test_idx) in enumerate(tscv.split(price)):
  train = price.iloc[train_idx]
  test = price.iloc[test_idx]

  #ищем лучший window на train данных
  best_window, train_results = find_best_window(train)

  #на test проверяем наш лучший window
  test_sharpe = run_strategy(test, best_window)

  wf_results.append({
      'fold': fold + 1,
      'best_window': best_window,
      'train_sharpe': train_results[best_window],
      'test_sharpe': test_sharpe
  })

  print(f"Фолд {fold + 1}:")
  print(f"  Лучший параметр на train: MA={best_window}")
  print(f"  Sharpe на train: {train_results[best_window]:.2f}")
  print(f"  Sharpe на test:  {test_sharpe:.2f}")
  print()

#выводим результат
results = pd.DataFrame(wf_results)
print("=" * 40)
print("ИТОГ WALK-FORWARD")
print("=" * 40)
print(results[['fold', 'best_window',
                   'train_sharpe', 'test_sharpe']])
print()
print(f"Средний Sharpe на train: {results['train_sharpe'].mean():.2f}")
print(f"Средний Sharpe на test:  {results['test_sharpe'].mean():.2f}")
```

Как читать результат:

Ситуация..................Train_Sharpe....Test_Sharpe....Вывод

Overfitting.................2.5......................0.1❌...............Стратегия запомнила прошлое

Рабочая стратегия..1.5.......................1.2✅..............Работает на новых данных

Наш пример..............0.03.................-0.55⚠️...........Есть overfitting, но стратегия что-то улавливает

# Деривативы

Claude написал простую историю пример, чтоб понять, что такое деривативы:

Сначала — простая история

Представь ты фермер. Сейчас март, пшеница стоит 100€ за тонну. Урожай будет в сентябре.

Ты думаешь: "А вдруг в сентябре цена упадёт до 50€? Я разорюсь!"

Покупатель хлеба думает: "А вдруг в сентябре цена вырастет до 200€? Я разорюсь!"

Они встречаются и договариваются прямо сейчас:

"В сентябре я продам тебе 100 тонн пшеницы по 100€ — что бы ни случилось с рынком"

Вот это и есть дериватив — контракт цена которого зависит от другого актива.

Существует 3 вида деривативов:

1. Фьючерсы (Futures)

    Многие знакомы с этим. Есть бессрочные и просто фьючерсы. Обычный фьючерс - это контракт, если заключил, отказаться нельзя(как на сво).
    Например: У тебя есть биткоин за 30к$, но ты думаешь что он упадет, поэтому заключаешь контракт на продажу биткоина через полгода за 30к.

    Проходит полгода и есть 2 варианта, либо упал, либо вырос, но ты в любом случае продаешь за 30к. Если биток стал 20к, ты в плюсе, если 40к, ты в минусе.

    Убыток или прибыль - разница в цене продажи и рынком

2. Опционы (Option)

    Если во фьючах, при заключении контракта, ты обязан испольнить свои обязательства, то в опиционах у тебя ПРАВО.

    Та же ситуация с битком по 30к$.
    
    Покупаешь премию (право), за условные 5$, продать биткоин за 30к$ через полгода.

    Проходит полгода и биток 40к, ты можешь не продавать и в итоге твой убыток только премия(5$), а если наоборот, биток 20к, то продаешь за 30к и прибыль 9995$.

3. Свопы (Swap)

    Это уже инструмент, которые обычные трейдеры не используют. Обычно это инструмент банков и корпораций.

    Своп - это буквально обмен, только обмениваются денежными потоками или обязательствами.

    Пример: ты взял кредит и у тебя не фиксированная оплата, а зависит от ставки ЦБ, а у друга такой же кредит, но с фиксированной оплатой на весь период.

    Вы договариваетесь, что будете платить друг за друга.

    В итоге у тебя фиксированная оплата, а у друга оплата зависит от ставки ЦБ

    НО своп это не право, а обязательство (контракт), если вы заключили такое соглашение, то вы обязаны его исполнять

# Типы торговых стратегий


Существует всего 2 стратегии: ***Trend Following и Mean Reversion***

Все остальное - это вариации этих двух стратегий

Дальше подробнее о каждой стратегии и также 2 вытекающих из них

*Trending Following* (следование тренду)

Это буквально следование за трендом. Логика такова: Если цена растет - она продолжит рост. Если падает - продолжит падать.

Плюсы:
*   Работает на сильных трендах
*   Простая логика
*   Используют крупные фонды

Минусы:

*   Много ложных сигналов
*   Плохо работает в боковике
*   Поздний вход и выход






*Mean Reversion* (возврат к среднему)

Логика стратегии такая: Все, что выросло - упадет обратно и все, что упало - вырастет обратно.

Плюсы:

*   Много сигналов
*   Работает в боковике
*   Четкий уровень входа

Минусы:

*   Актив может не вернуться к средней цене
*   Опасно против сильного тренда
*   Нужно угадать "среднюю цену"







> ⚠️ Главная опасность: акция упала на 30% — ты купил. Потом упала ещё на 50%. Mean reversion не сработал — это был просто плохой бизнес.



Дальше 2 стратегии, которые вытекают из двух предыдущих


*Momentum* (импульс)

Это стратегия вытекает из trending following. Суть: если актив рос 3-12 месяцев, то дальше он тоже будет рости

Разница между trending following в горизонте (у следования тренду это дни-недели, а у импульса это месяцы), на чем строится (следование тренду на технические индикаторы (MACD, MA), импульс - на прошлую доходность (график))

Чтоб лучше понять, пример:

Топ 10 акций (winners):   +40%, +35%, +32%...
Худшие 10 акций (losers): -30%, -25%, -20%...

Стратегия:
* Покупаешь топ 10 winners
* Продаёшь в шорт худшие 10 losers
* Держишь 1 месяц → переформируешь портфель





>🧠 Momentum — одна из самых исследованных аномалий рынка. Работает на акциях уже 100 лет. Никто точно не знает почему.



*Статистический арбитраж (Stat Arb)*

Стратегия строится на корреляции активов (2 актива имеют схожий график), т.е. условно ВТБ/СБЕР разница между ними примерно 50 пунктов всегда, если разница увеличилась (СБЕР вырос на 20 пунктов, а ВТБ нет), значит СБЕР можно брать в шорт, а ВТБ покупать (лонг)

# *VaR и CVaR* — сколько максимум можно потерять

**VaR - Value at Risk**

Это граница, т.е. у нас есть 252 торговых дня, я смотрю какая доходность была, сортирую от худших к лучшим и беру 5% от кол-во дней (252). Взяв этот процент, я получаю день, смотрю на какой день выпало по моему отсортированному списку и вижу, какой был минус (или плюс), значит, что в 95% случаев (дней) я не получал минус больше, чем в тот день и эта вероятность играет прямо здесь и сейчас, т.е. 95% того, что я не получу минус больше, чем на той границе.

Пример:

20 торговых дней:

+3%, +1%, -1%, +2%, -4%,
+0.5%, -2%, +1.5%, -6%, +2%,
+1%, -0.5%, +3%, -3%, +1%,
+2%, -1.5%, +0.5%, -8%, +1%

Сортирую:

-8%, -6%, -4%, -3%, -2%,
-1.5%, -1%, -0.5%, +0.5%, +0.5%,
+1%, +1%, +1%, +1.5%, +2%,
+2%, +2%, +3%, +3%, ...

Находим 5% от 20 дней: 20 * 5% = 1 день

Берем это значенпе: 1й худший день: -8%

VAR 95% = -8%

Т.е. 95% того, что я не получу сегодня минус больше чем -8%

**CVaR — Conditional Value at Risk**

Теперь, зная VAR мы можем посчитать CVAR. Он нужен для того, чтоб понять, какой средний минус в тех 5%

Складываем все доходности в днях, которые вышли за рамки VAR (например в 252 торговых дня это будет примерно 14 дней, значит смотрим в отсортированном списке худшие 14 дней)

Находим среднее - это и есть CVAR.

Т.е. теперь мы знаем, что если сегодня будет плохой день и мы перешагнем границу плохой доходности, то скорее всего она будет равна CVAR

#  Теория портфеля Марковица

Главная идея этой теории - **диверсификация портфеля**

Но главное не доходность активов, а то, как они коррелируют относительно друг друга

**Корреляция** - как активы двигаются вместе

+1.0 = двигаются всегда вместе

 0.0 = двигаются независимо

-1.0 = двигаются всегда противоположно

Почему важно учитывать корреляцию

Условно, если ты взял акции Apple и Microsoft, а они коррелируют +0.98

*Apple -3%, Microsoft -2.9%*

*Портфель: -2.95% ← оба упали вместе*

Но если бы ты взял Apple и Gold (у них корреляция -0.95), то ситуция выглядела бы так:

*Apple -3%, Золото +1.2%*
*Портфель: -0.9% ← золото смягчило падение!*



> 🧠 Вот почему диверсификация работает только когда активы не коррелированы. Купить Apple и Microsoft — это не диверсификация. Они падают и растут вместе.



***Efficient Frontier - граница эффективности***

Марковиц показал, что для **любого** уровня риска существует максимально возможная доходность

image.png

Сверху пример границы эффективности

Все, что до этой границы, имеет меньший риск, но и доходность меньше

Все, что дальше этой границы, имеют повышенный риск, за ту же доходность

3 портфеля, которые находятся на этой границе:

1.   **Maximum Sharpe Portfolio**

      Портфель с лучшим соотношением доходность/риск

2.   **Minimum Volatility Portfolio**

      Портфель с минимальным риском

3.   **Maximum Return Portfolio**

      Портфель с максимальной доходностью - но и максимальным риском





> Морковец доказал, что *больше доходность = больше риск*. Бесплатного сыра нет. НО с важным уточнением: Диверсификация позволяет снизить риск **БЕЗ** потери доходности - за счёт корреляции

Сейчас через PyPortfolioOpt автоматически найдем оптимальные веса для условного портфеля, состоящего из Apple, золота и облигаций

```python
import pandas as pd
import numpy as np
from pypfopt import EfficientFrontier
from pypfopt import risk_models, expected_returns
import matplotlib.pyplot as plt


#цена трех активов за 10 дней
prices = pd.DataFrame({
    'Apple':       [150, 152, 149, 155, 153, 158, 156, 160, 158, 163],
    'Gold':        [1800, 1798, 1805, 1795, 1802, 1798, 1806, 1800, 1808, 1804],
    'Bonds':       [100, 100.1, 100.2, 100.1, 100.3, 100.2, 100.4, 100.3, 100.5, 100.4]
})

#Ождиаемая годовая доходность каждого актива
mu = expected_returns.mean_historical_return(prices)
print(f"Ожидаемая доходность \n{mu.round(3)}")

#Ковариционная матрица - матрица корреляции (как активы двигаются вместе)
s = risk_models.sample_cov(prices)
print("\nКовариционная матрица")
print(s.round(6))

#Создаем объект Efficient Frontier
ef = EfficientFrontier(mu, s)

#Находим максимальный Sharpe
weights = ef.max_sharpe(risk_free_rate=0.04)

#Чистые веса
cleaned_weights = ef.clean_weights()
print("\nОптимальные веса (Max Sharpe):")
for asset, weight in cleaned_weights.items():
  print(f"  {asset}: {weight:.1%}")
print()

#Метрики портфеля
perfomance = ef.portfolio_performance(verbose=True)
```

```python
#Найдем минимальную волатильность портфеля

ef2 = EfficientFrontier(mu, s)
weight_minvol = ef2.min_volatility()
cleaned_weights2 = ef2.clean_weights()

print("\nОптимальные веса (Min Volatility):")
for asset, weight in cleaned_weights2.items():
    print(f"  {asset}: {weight:.1%}")
print()

ef2.portfolio_performance(verbose=True)
```

```python
#Рисуем Efficient Frontier

from pypfopt import  plotting

fig, ax = plt.subplots(figsize=(10,6))

#Рисуем границу эффективности
ef3 = EfficientFrontier(mu, s)
plotting.plot_efficient_frontier(ef3, asx=ax, show_assets=True)

#Отмечаем Max Sharpe
ef4 = EfficientFrontier(mu, s)
ef4.max_sharpe()
ret, vol, _ = ef4.portfolio_performance()
ax.scatter(vol, ret, marker='*', s=300,
           color='red', label='Max Sharpe', zorder=5)

#ret — доходность
#vol — риск
#_ — Sharpe (нам не нужен, поэтому просто _)

#Отмечаем Min Vol
ef5 = EfficientFrontier(mu, s)
ef5.min_volatility()
ret2, vol2, _ = ef5.portfolio_performance()
ax.scatter(vol2, ret2, marker='*', s=300,
           color='green', label='Min Volatility', zorder=5)

ax.set_title('Efficient Frontier')
ax.legend()
plt.show()
```

```python
#GARCH предсказывает завтрашнюю волатильность на основе прошлых движений

#сделаем синтетические данные

import numpy as np
import pandas as pd
from arch import arch_model
import matplotlib.pyplot as plt


np.random.seed(42)
returns = pd.Series(
    np.random.randn(500) * 0.01,
    name='returns'
)

# Масштабируем — arch любит числа в процентах а не долях
returns_pct = returns * 100

print(returns_pct.describe().round(3))
```

```python
# Создаём модель GARCH(1,1)
# p=1 — используем 1 прошлый квадрат доходности
# q=1 — используем 1 прошлую дисперсию

model = arch_model(returns_pct, #тип модели
                   vol='Garch',
                   p=1, #порядок ARCH
                   q=1, #порядок GARCH
                   dist='normal') #распределение остатков

print(model)
```



> 🧠 GARCH(1,1) — самая популярная версия. Цифры в скобках означают:
>*   p=1 → смотрим на 1 прошлый день доходностей
>*   q=1 → смотрим на 1 прошлый день волатильности




```python
#Подбираем параметры на данных
result = model.fit(disp='off') #disp='off' — не печатать лог
print(result.summary())
```

Что значат параметры:

**mu** - среднее доходностей

**omega** - базовый уровень волатильности

**alpha[1]** - вес вчерашней доходности

**beta[1]** - вес вчерашней волатильности



> 🎯 beta[1] = 0.99. Это очень высокое значение — значит волатильность очень инертна. Если вчера было бурно — сегодня тоже скорее всего бурно. Это и есть кластеризация.



```python
# Условная волатильность — что модель думала о волатильности каждый день

cond_vol = result.conditional_volatility

plt.figure(figsize=(12, 5))
plt.plot(cond_vol, color='blue', linewidth=1)
plt.title('Условная волатильность (GARCH)')
plt.ylabel('Волатильность (%)')
plt.xlabel('День')
plt.axhline(cond_vol.mean(), color='red',
            linestyle='--', label=f'Среднее: {cond_vol.mean():.2f}%')
plt.legend()
plt.show()

print(f"Средняя волатильность: {cond_vol.mean():.3f}%")
print(f"Макс волатильность:    {cond_vol.max():.3f}%")
print(f"Мин волатильность:     {cond_vol.min():.3f}%")
```

```python
#предсказываем будущую волатильность

forecast = result.forecast(horizon=5)

variance_forecast = forecast.variance.iloc[-1]
vol_forecast = np.sqrt(variance_forecast)

print("Прогноз волатильности:")
for day, vol in enumerate(vol_forecast, 1):
  print(f"День +{day}: {vol:.3f}%")
```

```python
# Идея: уменьшаем размер позиции когда волатильность высокая
avg_vol = cond_vol.mean()

# Размер позиции обратно пропорционален волатильности
position_size = avg_vol / cond_vol

print("Примеры размера позиции:")
print(f"  Спокойный день (vol=0.89%):  {avg_vol/0.89:.2f}x")
print(f"  Обычный день   (vol=0.98%):  {avg_vol/0.98:.2f}x")
print(f"  Бурный день    (vol=1.31%):  {avg_vol/1.31:.2f}x")
```



> 🎯 В бурный день — уменьшаем позицию до 75%. В спокойный — увеличиваем до 110%. Это динамический риск менеджмент через GARCH.

    **result = model.fit(disp='off')**

# Feature Engineering

Если мы хотим научить ML модель предсказывать цену, мы не можем просто скормить ей цену.

Мы должны дать ей разные признаки (Доходность вчера, Импульс за 20д, Средняя доходность 5д) - это и есть ***Feature Engineering***

*ТРИ ГРУППЫ ФИНАНСОВЫХ ФИЧ*

```python
#Доходности - первая фича

import pandas as pd
import numpy as np

#Данные цены за 300 дней
np.random.seed(42)
prices = pd.Series(
    100 * (1 + np.random.randn(300) * 0.01).cumprod(),
    name='price'
)

returns = prices.pct_change()

#Лаговые доходности — доходность 1, 2, 3 дня назад
lag1 = returns.shift(1)
lag2 = returns.shift(2)
lag3 = returns.shift(3)

print(pd.DataFrame({
    'price':  prices,
    'return': returns,
    'lag1':   lag1,
    'lag2':   lag2,
    'lag3':   lag3
}).round(4).head(8))
```



> 🧠 Лаги — это память модели. Ты говоришь ей: "Вот что происходило 1, 2, 3 дня назад — используй это для предсказания"



```python
#Rolling Volatility (скользящая волатильность) - вторая фича

vol_5 = returns.rolling(5).std()
vol_10 = returns.rolling(10).std()
vol_20 = returns.rolling(20).std()

print(pd.DataFrame({
    'return': returns,
    'vol_5':  vol_5,
    'vol_10': vol_10,
    'vol_20': vol_20
}).round(4).tail(8))
```



> 🎯 Три временных горизонта дают модели понимание — рынок сейчас спокойный или бурный? Краткосрочно или долгосрочно?



```python
#Momentum (импульс) - третья фича

#Импульс - доходность за последние N дней
mom_5 = prices.pct_change(5)
mom_10 = prices.pct_change(10)
mom_20 = prices.pct_change(20)


#Скользящее среднее
ma_5 = prices.rolling(5).mean()
ma_10 = prices.rolling(10).mean()
ma_20 = prices.rolling(20).mean()

#Расстояние цены от скользящей средней
dist_ma5  = (prices - ma_5)  / ma_5   #насколько выше/ниже MA5
dist_ma20 = (prices - ma_20) / ma_20  #насколько выше/ниже MA20

print(pd.DataFrame({
    'mom_5':     mom_5,
    'mom_20':    mom_20,
    'dist_ma5':  dist_ma5,
    'dist_ma20': dist_ma20
}).round(4).tail(5))
```



> 🧠 dist_ma5 = -0.008 значит цена сейчас на 0.8% ниже своей 5-дневной средней. Это сигнал краткосрочного нисходящего импульса.



Теперь собираем все воедино и главное, фичи должны быть сдвинуты на 1 день - иначе **look-ahead bias** (видим будущее)

```python
features = pd.DataFrame({
  'return_lag1': returns.shift(1),
  'return_lag2': returns.shift(2),
  'return_lag3': returns.shift(3),

  'vol_5':  vol_5.shift(1),
  'vol_10': vol_10.shift(1),

  'mom_5':     mom_5.shift(1),
  'mom_20':    mom_20.shift(1),
  'dist_ma5':  dist_ma5.shift(1),
  'dist_ma20': dist_ma20.shift(1),
})

#Таргет - вырастет ли цена завтра?
#1 = вырастет, 0 = упадёт
target = (returns > 0).astype(int)

#Чистим строки с NaN
features = features.dropna()
target = target.loc[features.index]

print(f"Фичей: {features.shape[1]}")
print(f"Строк: {features.shape[0]}")
print(features.round(4).tail(5))
```

Теперь фичи кидаем в CatBoost

```python
pip install catboost
```

```python
from catboost import CatBoostClassifier
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import accuracy_score
import numpy as np

X = features
y = target

#walk-forward validation
tscv = TimeSeriesSplit(n_splits=3)
scores = []

for fold, (train_idx, test_idx) in enumerate (tscv.split(X)):
  X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
  X_test, y_test = X.iloc[test_idx], y.iloc[test_idx]

  #Модель

  model = CatBoostClassifier( iterations = 100, learning_rate = 0.1, depth=4, verbose = 0)
  model.fit(X_train, y_train)
  pred = model.predict(X_test)

  acc = accuracy_score(y_test, pred)
  scores.append(acc)
  print(f"Фолд {fold+1}: accuracy = {acc:.2%}")


print(f"\nСредняя accuracy: {np.mean(scores):.2%}")
```
